## 7 Filter rows
(a) Make a mask for listings with two or more bedrooms. Use sum() and mean() on it: how many are there, and what share of all listings?
(b) Find the listings that are in Östermalms, are Entire home/apt and have two or more bedrooms. 
Start with and between the conditions and read the error. Then fix it with & and parentheses, and print the shape.

On purpose: Use and first, and read the error from the bottom up.

```python
# 7(a) Make a mask for listings with two or more bedrooms (>= 2)

bedrooms_mask = listings['bedrooms'] >= 2

count_bedrooms = bedrooms_mask.sum()
share_bedrooms = bedrooms_mask.mean() * 100

print(f"--- Listings with 2 or more bedrooms ---\n")
print(f"Count: {count_bedrooms}")
print(f"Share: {share_bedrooms:.2f}%\n")


# 7(b) On purpose: Trying to use 'and' between conditions (this will trigger a ValueError)

# error_test = listings[(listings['neighbourhood_cleansed'] == 'Östermalms') and (listings['room_type'] == 'Entire home/apt')]

# Correct version using '&' and parentheses
correct_mask = (
    (listings['neighbourhood_cleansed'] == 'Östermalms') & 
    (listings['room_type'] == 'Entire home/apt') & 
    (listings['bedrooms'] >= 2)
)

filtered_listings = listings[correct_mask]

print("--- Filtered listings shape (Östermalms, Entire home/apt, 2+ bedrooms) ---")
print(filtered_listings.shape)
```

## 8 Sort and pick
Find the five listings in Östermalms with the most reviews. 
Show name, room_type and number_of_reviews, with the biggest number of reviews first. 
Which room type does the listing at the top have?

```python
# 8. Find the top 5 listings in Östermalms with the most reviews

ostermalm_mask = listings['neighbourhood_cleansed'] == 'Östermalms'
ostermalm_listings = listings[ostermalm_mask]


cols_to_show = ['name', 'room_type', 'number_of_reviews']

# Sort by number_of_reviews descending and pick the top 5
top_reviewed_ostermalm = ostermalm_listings[cols_to_show].sort_values(
    by='number_of_reviews', 
    ascending=False
).head(5)

print("--- Top 5 most reviewed listings in Östermalms ---\n")
print(top_reviewed_ostermalm)
```

## 2.1 Central or not
Make a list central with the four central neighbourhoods Södermalms, Östermalms, Norrmalms and Kungsholmens, and make a mask with isin(). 
How many listings are central, and what share is that of all listings?
Then compare the share of private rooms among the central listings with the share outside the centre (use ~). 
Is it higher in the centre or outside?

```python
# Part 2 - Task 1: Central or not

# Define the list of central neighbourhoods and create a mask using .isin()
central = ['Södermalms', 'Östermalms', 'Norrmalms', 'Kungsholmens']
central_mask = listings['neighbourhood_cleansed'].isin(central)

# Count how many listings are central and calculate their share (%)
n_central = central_mask.sum()
share_central = central_mask.mean() * 100

print(f"Number of central listings: {n_central}")
print(f"Share of central listings: {share_central:.2f}%\n")

# Compare the share of private rooms in the centre vs. outside the centre
# Create a mask for private rooms
private_room_mask = listings['room_type'] == 'Private room'

# Calculate share of private rooms among central listings
share_private_central = private_room_mask[central_mask].mean() * 100

# Calculate share of private rooms outside the centre
share_private_outside = private_room_mask[~central_mask].mean() * 100

print(f"Share of private rooms in the centre: {share_private_central:.2f}%")
print(f"Share of private rooms outside the centre: {share_private_outside:.2f}%")
```

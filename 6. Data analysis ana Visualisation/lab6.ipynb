** 1.1 The mean and the median **

Load the table. First, make a small pd.
Series with five numbers of reviews that are close to each other, and calculate the mean and the median. 
Then add one listing with 1293 reviews (the most popular one) and calculate again. 
What happened to each number?
Then calculate the mean and the median of the real column number_of_reviews for all 4,574 listings. 
Which one would you use to answer "how many reviews does a typical listing have?", and why?

```python
import pandas as pd

# Load the table:
listings = pd.read_csv('data/listings_clean.csv')

# Create a small Series of 5 review:
reviews_small = pd.Series([3, 5, 8, 10, 12])
print("Small list - Mean:", reviews_small.mean())
print("Small list - Median:", reviews_small.median())

# Add one extreme outlier (1293 reviews) and recalculate mean and median
reviews_with_outlier = pd.Series([3, 5, 8, 10, 12, 1293])
print("\nWith 1293 added - Mean:", round(reviews_with_outlier.mean(), 2))
print("With 1293 added - Median:", reviews_with_outlier.median())

# Calculate the mean and median for the actual number_of_reviews column across all listings
real_mean = listings['number_of_reviews'].mean()
real_median = listings['number_of_reviews'].median()

print(f"\nAll listings ({len(listings)} rows):")
print(f"Mean number of reviews: {real_mean:.2f}")
print(f"Median number of reviews: {real_median:.2f}")
```

** 1.2 Percentiles and the spread **

Run describe() on number_of_reviews. 
Then calculate the 10th, 25th, 50th, 75th, 90th and 99th percentile with quantile. 
Calculate q1, q3 and the iqr. Write one sentence in your own words about what the IQR says here. 
Look at the 10th percentile: what does it tell you about how many listings have no reviews at all?

```python
# Run describe() on number_of_reviews. 
print(listings['number_of_reviews'].describe())

# Then calculate the 10th, 25th, 50th, 75th, 90th and 99th percentile with quantile. 
percentiles = listings['number_of_reviews'].quantile([0.1, 0.25, 0.5, 0.75, 0.9, 0.99])
print("\nPercentiles:")
print(percentiles)

# Calculate q1, q3 and the iqr - (Interquartile Range)
q1 = listings['number_of_reviews'].quantile(0.25)
q3 = listings['number_of_reviews'].quantile(0.75)
iqr = q3 - q1

print(f"\nQ1 (25th percentile): {q1}")
print(f"Q3 (75th percentile): {q3}")
print(f"IQR: {iqr}")

#The 10th percentile is 0 (and since the number of reviews cannot be negative, the minimum is also 0).
```

** 1.3 The fences and the outliers **

Calculate the upper and the lower fence with the IQR rule. 
Can there be outliers below the lower fence for this column? 
Why, or why not? How many listings are above the upper fence? 
Then look at them: how many of each room type, and what is their 
median price compared with all the other listings? 
Write what you see in a comment.

```python
# Calculate the upper and lower fences (using the 1.5 * IQR rule...)
q1 = listings['number_of_reviews'].quantile(0.25)
q3 = listings['number_of_reviews'].quantile(0.75)
iqr = q3 - q1

upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

print(f"Lower fence: {lower}")
print(f"Upper fence: {upper}")

# Mark the review outliers
listings['reviews_outlier'] = listings['number_of_reviews'] > upper

# Count how many listings are above the upper fence
outliers = listings[listings['reviews_outlier']]
non_outliers = listings[~listings['reviews_outlier']]  # kontra

print(f"\nNumber of listings above upper fence: {len(outliers)}")

# Check how many of each room_type are among the outliers
print("\nRoom types among high-review outliers:")
print(outliers['room_type'].value_counts())

# Compare median price of high-review outliers vs. all other listings
print(f"\nMedian price of outliers: {outliers['price'].median()}")
print(f"Median price of all other listings: {non_outliers['price'].median()}")

# Findings from the fences and outliers:
# - There cannot be outliers below the lower fence (-38.5) because a listing cannot have negative reviews (the minimum is 0).
# - There are 582 listings above the upper fence (69.5 reviews), 
# - mostly Entire home/apt (446) and Private room (121).
# - High-review outliers are noticeably cheaper: 
#   their median price is 1,459.54 SEK compared to 1,867.00 SEK for all other listings.
```

** 1.4 Mark them, and compare **

Do not remove the outliers. 
Mark them in a new column, reviews_outlier (True or False), and count them. 
Then make a table with pd.DataFrame that shows count, mean, median and
std for number_of_reviews, with all listings and without the outliers. 
How many percent does each number change when the outliers are gone? 
Which changes the most, and which the least?

```python
# Do not remove the outliers:
listings['reviews_outlier'] = listings['number_of_reviews'] > upper

# Count True vs False
print("Outlier counts:")
print(listings['reviews_outlier'].value_counts())

# Create series for all listings vs. without outliers
all_reviews = listings['number_of_reviews']
without_outliers = listings[~listings['reviews_outlier']]['number_of_reviews']

# Build a comparison DataFrame with count, mean, median, and std
stats = ['count', 'mean', 'median', 'std']
comparison = pd.DataFrame({
    'all': all_reviews.agg(stats),
    'without': without_outliers.agg(stats)
})

# Calculate percentage change when outliers are removed
comparison['pct_change'] = ((comparison['without'] - comparison['all']) / comparison['all']) * 100

print("\nComparison table (All vs. Without Outliers):")
print(comparison.round(2))
```

** 1.6 Correlation **

Calculate the correlation between number_of_reviews and these columns:

```python
# Select the columns to check correlation with number_of_reviews
cols = ['number_of_reviews', 'price', 'accommodates', 'bedrooms', 'availability_365', 'review_scores_rating', 'minimum_nights', 'days_since_review']

# Calculate correlation with number_of_reviews, rounded to 2 decimals
corrs = listings[cols].corr()['number_of_reviews'].round(2)
print("Correlations with number_of_reviews (all listings):")
print(corrs.sort_values(ascending=False))

# Compare correlation between number_of_reviews and availability_365:
# with all listings vs. without the review outliers from Task 4
corr_all = listings['number_of_reviews'].corr(listings['availability_365']).round(2)

without_outliers_df = listings[~listings['reviews_outlier']]
corr_no_outliers = without_outliers_df['number_of_reviews'].corr(without_outliers_df['availability_365']).round(2)

print(f"\nCorrelation with availability_365 (all listings): {corr_all}")
print(f"Correlation with availability_365 (without outliers): {corr_no_outliers}")
```

# What happens, and what does it tell you about the first number?

## 1.6 - 1. Summary of All Correlations (number_of_reviews vs. other columns)
- Strongest connection: days_since_review (-0.19), followed closely by availability_365 (0.18) — though even the strongest correlations here are fairly weak in absolute terms.

- Weakest connection: review_scores_rating (-0.04), showing almost zero linear relationship between a listing's star rating and how many reviews it has.

- Negative correlations: days_since_review (-0.19), bedrooms (-0.14), price (-0.09), minimum_nights (-0.07), accommodates (-0.05), and review_scores_rating (-0.04). For example, listings with more reviews tend to have been reviewed more recently (days_since_review is smaller) and tend to be smaller/cheaper properties (bedrooms and price are lower).

## 1.6 - 2. Correlation with availability_365: With vs. Without Outliers

- What happens: The correlation drops completely from 0.18 down to 0.00 once the 582 review outliers are removed.

- What it tells us about the first number (0.18): The initial positive correlation of 0.18 was entirely driven by the extreme outliers (the small group of commercial/full-time listings that stay open year-round and have hundreds of reviews). Among the remaining ~87% of typical listings (without outliers), there is zero linear relationship (0.00) between how many days a listing is available and how many reviews it receives. This is a textbook example of how Pearson correlation can be misleading when a dataset has heavy outliers.

## 1 Count property types
Load the file. Use value_counts() on property_type and show only the five most common
types. What share of all listings is Entire rental unit? (Use normalize=True, rounded to
three decimals.)

```python
import pandas as pd
import numpy as np

# Problem with .csv download...
listings = pd.read_csv('data/listings.csv')

# 1. Top 5
top5_types = listings['property_type'].value_counts().head(5)
print("--- Top 5 property types ---")
print(top5_types)

# 2. Shares
shares = listings['property_type'].value_counts(normalize=True)

# Calculate 'Entire rental unit'
entire_share = shares.get('Entire rental unit', 0)
print(f"\nShare for 'Entire rental unit': {entire_share:.3f}")
```

Use value_counts() on neighbourhood_cleansed. Which three neighbourhoods have the

## 3 Average beds per room type
Calculate the average of beds for each room_type, rounded to two decimals. Then use
agg(['count', 'mean']) on the same thing. Which average do you trust the least, and why?

```python
avg_beds = listings.groupby('room_type')['beds'].mean().round(2)
print("--- Average number of beds per room type ---\n")
print(avg_beds)

agg_beds = listings.groupby('room_type')['beds'].agg(['count', 'mean'])
print("\n--- Detailed aggregation (count and mean) ---")
print(agg_beds)
```

## 4 Reviews per neighbourhood
Group by neighbourhood_cleansed and calculate count, mean and median of
number_of_reviews, rounded to two decimals, sorted by the median. 
Which neighbourhood has the lowest median and which the highest? 
For Södermalms the mean is much bigger than the median. What could be the reason?

The mean is much higher than the median for Södermalms because number of reviews is typically right-skewed. 
Södermalm is a very popular and central area with a large number of long-standing and hundreds of reviews...

```python
reviews_stats = listings.groupby('neighbourhood_cleansed')['number_of_reviews'].agg(['count', 'mean', 'median']).round(2)

# Sort by median in ascending order
reviews_stats_sorted = reviews_stats.sort_values(by='median')

print("--- Neighbourhood review statistics (sorted by median) ---")
print(reviews_stats_sorted)

# Display the neighbourhood with the lowest and highest median
lowest_median = reviews_stats_sorted.index[0]
highest_median = reviews_stats_sorted.index[-1]

print(f"\nNeighbourhood with the lowest median: {lowest_median}")
print(f"Neighbourhood with the highest median: {highest_median}")
```

## 5 Your own summary table
Make a summary per room_type with named columns: n_listings (count of id), avg_beds
(mean of beds) and max_reviews (max of number_of_reviews). Round to two decimals.

```python
summary_table = listings.groupby('room_type').agg(
    n_listings=('id', 'count'),
    avg_beds=('beds', 'mean'),
    max_reviews=('number_of_reviews', 'max')
).round(2)

print("          --- Summary table per room type --- ")
print(summary_table)
```

## 6 Two categories at once
Make a crosstab of neighbourhood_cleansed and host_is_superhost with the share in
percent for each neighbourhood (normalize='index', times 100, one decimal). The column
t means that the host is a superhost. Which neighbourhood has the highest share of
superhosts, and which the lowest?

```python
superhost_share = pd.crosstab(
    listings['neighbourhood_cleansed'], 
    listings['host_is_superhost'], 
    normalize='index'
) * 100

superhost_share = superhost_share.round(1)

print("--- Superhost share per neighbourhood (%) ---\n")
print(superhost_share)

# Find neighbourhood with the highest and lowest share of superhosts:
highest_superhost = superhost_share['t'].idxmax()
lowest_superhost = superhost_share['t'].idxmin()

print(f"\nNeighbourhood with the highest share of superhosts: {highest_superhost} ({superhost_share.loc[highest_superhost, 't']}%)")
print(f"Neighbourhood with the lowest share of superhosts: {lowest_superhost} ({superhost_share.loc[lowest_superhost, 't']}%)")
```

## 7 Create new columns
(a) Create a column has_reviews that is True when number_of_reviews is bigger than 0.
How many listings are False?
(b) Create a column popularity with np.where: 'popular' if the listing has 50 reviews or
more, otherwise 'other'. Count the values.

```python
import numpy as np

# 7(a) Create a column 'has_reviews' that is True when number_of_reviews is greater than 0
listings['has_reviews'] = listings['number_of_reviews'] > 0

print("    --- Count of True/False for has_reviews ---\n")
print(listings['has_reviews'].value_counts())

# 7(b) Create a column 'popularity' using np.where: 'popular' if >= 50 reviews, otherwise 'other'
listings['popularity'] = np.where(listings['number_of_reviews'] >= 50, 'popular', 'other')

print("\n    --- Value counts for popularity column ---\n")
print(listings['popularity'].value_counts())
```

## 8 Turn numbers into groups
The column minimum_nights says how long you have to stay. Make four groups with pd.cut:
1 night, 2-6 nights, 7-29 nights and 30 or more nights. Save them as a column stay with
labels, and count the groups with value_counts().sort_index(). How many listings have no
group?
## On purpose: First run the cut with bins=[1, 6, 29, 500] and count the missing values with
isna().sum(). Why are so many missing? Then fix it.

```python
import pandas as pd
import numpy as np

# 8(a) Test run with bins=[1, 6, 29, 500] on purpose:
bins_test = [1, 6, 29, 500]
listings['stay_test'] = pd.cut(listings['minimum_nights'], bins=bins_test)

missing_count = listings['stay_test'].isna().sum()
print(f"Number of missing (NaN) values with test bins: {missing_count}")

# 8(b) Fixed version: 4 groups (1 night, 2-6 nights, 7-29 nights, 30+ nights)
bins_correct = [0, 1, 6, 29, float('inf')]
labels_correct = ['1 night', '2-6 nights', '7-29 nights', '30+ nights']

listings['stay'] = pd.cut(listings['minimum_nights'], bins=bins_correct, labels=labels_correct)

stay_counts = listings['stay'].value_counts().sort_index()
print("\n--- Stay groups count ---\n")
print(stay_counts)

missing_correct = listings['stay'].isna().sum()
print(f"Listings with no group (NaN) after fix: {missing_correct}")
```

## 2.1 Who has never been reviewed?
Use the column has_reviews from Part 1 (or create it again). 
What share of all listings has no reviews? 
Then calculate the share without reviews in each neighbourhood. 
Which neighbourhoods have the lowest and the highest share? 

(Hint: the mean of True/False is a share, so think about which of the two you want the mean of.)

```python
# Part 2 - Task 1: Who has never been reviewed?

# Make sure 'has_reviews' column exists (True if number_of_reviews > 0)
listings['has_reviews'] = listings['number_of_reviews'] > 0

# Create a boolean mask for listings with NO reviews
listings['no_reviews'] = ~listings['has_reviews']

# Calculate total share of listings with no reviews
overall_no_reviews_share = listings['no_reviews'].mean() * 100
print(f"Overall share of listings with no reviews: {overall_no_reviews_share:.2f}%\n")

# Calculate share without reviews per neighbourhood
neighbourhood_no_reviews = listings.groupby('neighbourhood_cleansed').agg(
    n_listings=('id', 'count'),
    share_no_reviews=('no_reviews', 'mean')
)

# Convert share to percentage and round to 2 decimals
neighbourhood_no_reviews['share_no_reviews_pct'] = (neighbourhood_no_reviews['share_no_reviews'] * 100).round(2)

# Sort values to find lowest and highest shares
neighbourhood_sorted = neighbourhood_no_reviews.sort_values(by='share_no_reviews_pct')

print("--- Share of listings without reviews per neighbourhood (%) ---")
print(neighbourhood_sorted[['n_listings', 'share_no_reviews_pct']])

# Identify lowest and highest
lowest_share_nh = neighbourhood_sorted.index[0]
highest_share_nh = neighbourhood_sorted.index[-1]

print(f"\nNeighbourhood with the lowest share of unreviewed listings: {lowest_share_nh} ({neighbourhood_sorted.loc[lowest_share_nh, 'share_no_reviews_pct']}%)")
print(f"Neighbourhood with the highest share of unreviewed listings: {highest_share_nh} ({neighbourhood_sorted.loc[highest_share_nh, 'share_no_reviews_pct']}%)")
```
